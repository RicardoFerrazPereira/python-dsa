```python
# Criando uma classe

class Funcionarios:
    
    def __init__(self, nome, salario, cargo):
        self.nome = nome
        self.salario = salario
        self.cargo = cargo

    def listFunc(self):
        print("Funcionário(a) " + self.nome + " tem salário de R$" + str(self.salario) + " e o cargo é " + self.cargo)
```

```python
# Criando um objeto chamado Func1 a partir da classe Funcionarios
Func1 = Funcionarios("Eddie", 50000, "Vagabundo")
```

```python
Func1.listFunc()
```

```python
print("**** Usando atributos *****")
```

```python
hasattr(Func1, "nome") # hasattr -> tem atributo nome em Func1?
```

```python
hasattr(Func1, "salario")
```

```python
setattr(Func1, "salario", 4500) # setattr -> alterar atributo -> mude o salarios do func1 para 4500
```

```python
getattr(Func1, "salario") # pegue o valor do salário do objeto func1
```
