## Pergunta de Negócio 5:

### Qual Segmento Teve o Maior Total de Vendas?

Demonstre o resultado através de um gráfico de pizza.

```python
df_p5 = df.groupby('Segmento')['Valor_Venda'].sum().sort_values(ascending=False).reset_index()
df_p5.head()
```

```python
# Função para converter os dados em valor absoluto
def autopct_format(values): 
    def my_format(pct): 
        total = sum(values) 
        val = int(round(pct * total / 100.0))
        return ' $ {v:d}'.format(v = val)
    return my_format
```

```python
# Plot

# Tamanho da figura
plt.figure(figsize = (16, 6))

# Gráfico de pizza
plt.pie(df_p5['Valor_Venda'], 
        labels = df_p5['Segmento'],
        autopct = autopct_format(df_p5['Valor_Venda']),
        startangle = 90)

# Limpa o círculo central
centre_circle = plt.Circle((0, 0), 0.82, fc = 'white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

# Labels e anotações
plt.annotate(text = 'Total de Vendas: ' + '$ ' + str(int(sum(df_p5['Valor_Venda']))), xy = (-0.25, 0))
plt.title('Total de Vendas Por Segmento')
plt.show()
```

## Pergunta de Negócio 6 (Desafio Nível Baby):

### Qual o Total de Vendas Por Segmento e Por Ano?

```python
# Converter Data_Pedido para datetime
df['Data_Pedido'] = pd.to_datetime(df['Data_Pedido'])
```

```python
df.dtypes
```

```python
# Criar a coluna de Ano
df['Ano'] = df['Data_Pedido'].dt.year

# Agrupar por Segmento e Ano e calcular o total de vendas
total_vendas = df.groupby(['Segmento', 'Ano'])['Valor_Venda'].sum().reset_index()

# Ordenar os resultados (opcional)
total_vendas = total_vendas.sort_values(['Ano', 'Valor_Venda'], ascending=[True, False])

print(total_vendas)
```

```python
total_vendas = df.groupby(['Segmento', 'Ano']).agg(
    Total_Vendas=('Valor_Venda', 'sum'),
    Qtde_Pedidos=('ID_Pedido', 'nunique')  # Adicionando contagem de pedidos únicos
).reset_index()

# Ordenação por Ano (crescente) e Total_Vendas (decrescente)
vendas_ordenadas = total_vendas.sort_values(
    by=['Ano', 'Total_Vendas'],
    ascending=[True, False])
print(vendas_ordenadas)
```

## Pergunta de Negócio 7 (Desafio Nível Júnior):

Os gestores da empresa estão considerando conceder diferentes faixas de descontos e gostariam de fazer uma simulação com base na regra abaixo:

- Se o Valor_Venda for maior que 1000 recebe 15% de desconto.
- Se o Valor_Venda for menor que 1000 recebe 10% de desconto.

### Quantas Vendas Receberiam 15% de Desconto?

```python
# Aplicando a regra de desconto e calculando o valor do desconto
df['desconto'] = df['Valor_Venda'].apply(lambda x: x * 0.15 if x > 1000 else x * 10)

# Contando quantas vendas receberam 15% de desconto
qtd_15_desconto = df[df['Valor_Venda'] > 1000].shape[0]

# Contando quantas vendas receberam 10% de desconto
qtd_10_desconto = df[df['Valor_Venda'] < 1000].shape[0]

df
```

```python
# Contando quantas vendas receberam 10% e 15% de desconto
print(f"\nQuantidade de vendas que receberam 10% de desconto: {qtd_10_desconto}")
print(f"\nQuantidade de vendas que receberam 15% de desconto: {qtd_15_desconto}")
```

## Pergunta de Negócio 8 (Desafio Nível Master):

### Considere Que a Empresa Decida Conceder o Desconto de 15% do Item Anterior. Qual Seria a Média do Valor de Venda Antes e Depois do Desconto?

```python
# Aplicando a regra de desconto e calculando o valor do desconto
df['desconto'] = df['Valor_Venda'].apply(lambda x: x * 0.15 if x > 1000 else x * 0.10)

# Calculando o valor de venda após o desconto
df['valor_final'] = df['Valor_Venda'] - df['desconto']

# Calculando as médias antes e depois do desconto
media_antes = df['Valor_Venda'].mean()
media_depois = df['valor_final'].mean()

df
```

```python
print(f"\nMédia do valor de venda antes do desconto: {media_antes:.2f}")
print(f"Média do valor de venda depois do desconto: {media_depois:.2f}")
```

## Pergunta de Negócio 9 (Desafio Nível Master Ninja):

### Qual o Média de Vendas Por Segmento, Por Ano e Por Mês?

Demonstre o resultado através de gráfico de linha.

```python
# Criar a coluna de Mês
df['Mes'] = df['Data_Pedido'].dt.month

df.head()
```

```python
df = df.drop(columns=['mes'])

df.head()
```

```python
# Agrupar por Segmento e Ano e calcular o total de vendas
media_vendas = df.groupby([ 'Ano', 'Mes', 'Segmento'])['Valor_Venda'].agg([np.sum, np.mean, np.median])


media_vendas
```

```python
# Vamos extrair os níveis
anos = media_vendas.index.get_level_values(0)
meses = media_vendas.index.get_level_values(1)
segmentos = media_vendas.index.get_level_values(2)
```

```python
# Plot
plt.figure(figsize = (12, 6))
sns.set()
fig1 = sns.relplot(kind = 'line',
                   data = media_vendas, 
                   y = 'mean', 
                   x = meses,
                   hue = segmentos, 
                   col = anos,
                   col_wrap = 4)
plt.show()
```

## Pergunta de Negócio 10 (Desafio Nível Master Ninja das Galáxias):

### Qual o Total de Vendas Por Categoria e SubCategoria, Considerando Somente as Top 12 SubCategorias? 

Demonstre tudo através de um único gráfico.

```python
df_total = df.groupby(['Categoria', 'SubCategoria']).sum(numeric_only = True).sort_values('Valor_Venda',
                                                                                   ascending = False).head(12)

df_total
```

```python
# Convertemos a coluna Valor_Venda em número inteiro e classificamos por categoria
df_total = df_total[['Valor_Venda']].astype(int).sort_values(by = 'Categoria').reset_index()
```

```python
# Dataframe com categorias e subcategorias
df_total
```

```python
# Criamos outro dataframe somente com os totais por categoria
totais_categoria = df_total.groupby('Categoria').sum(numeric_only = True).reset_index()
```

```python
# Dataframe com categorias 
totais_categoria
```

```python
# Listas de cores para categorias
cores_categorias = ['#5d00de',
                    '#0ee84f',
                    '#e80e27']
```

```python
# Listas de cores para subcategorias
cores_subcategorias = ['#aa8cd4',
                       '#aa8cd5',
                       '#aa8cd6',
                       '#aa8cd7',
                       '#26c957',
                       '#26c958',
                       '#26c959',
                       '#26c960',
                       '#e65e65',
                       '#e65e66',
                       '#e65e67',
                       '#e65e68']
```

```python
# Plot

# Tamanho da figura
fig, ax = plt.subplots(figsize = (18,12))

# Gráfico das categorias
p1 = ax.pie(totais_categoria['Valor_Venda'], 
            radius = 1,
            labels = totais_categoria['Categoria'],
            wedgeprops = dict(edgecolor = 'white'),
            colors = cores_categorias)

# Gráfico das subcategorias
p2 = ax.pie(df_total['Valor_Venda'],
            radius = 0.9,
            labels = df_total['SubCategoria'],
            autopct = autopct_format(df_total['Valor_Venda']),
            colors = cores_subcategorias, 
            labeldistance = 0.7,
            wedgeprops = dict(edgecolor = 'white'), 
            pctdistance = 0.53,
            rotatelabels = True)

# Limpa o centro do círculo
centre_circle = plt.Circle((0, 0), 0.6, fc = 'white')

# Labels e anotações
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.annotate(text = 'Total de Vendas: ' + '$ ' + str(int(sum(df_total['Valor_Venda']))), xy = (-0.2, 0))
plt.title('Total de Vendas Por Categoria e Top 12 SubCategorias')
plt.show()
```
